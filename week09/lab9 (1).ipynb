## Part 1: NumPy

NumPy is a Python library used for working with arrays and performing numerical calculations.

Practice Question: The Plant Growth Experiment

You are studying the growth of six plants over a week. Their growth measurements (in centimetres) are:

`[4, 7, 5, 9, 6, 8]`

- Create a NumPy array called `growth`.

- Calculate the average growth.

- Find the maximum and minimum growth.

- Calculate the total growth across all six plants.


````python
growth = np.array([4, 7, 5, 9, 6, 8])

print("Average growth:", np.mean(growth))
print("Maximum growth:", np.max(growth))
print("Minimum growth:", np.min(growth))
print("Total growth:", np.sum(growth))
````

### Task 1
Task: Temperature Monitoring

A scientist records the temperature inside a greenhouse over five days.

The recorded temperatures are:

`[22, 25, 21, 28, 24]`

Using NumPy:

- Create an array called `temperatures`.

- Calculate the average temperature.

- Find the highest and lowest temperatures.

- Find all temperatures above 23°C using Boolean filtering.

- Count how many days had temperatures above 23°C.

````python
temperatures = np.array([22, 25, 21, 28, 24])

average_temp = np.mean(temperatures)
highest_temp = np.max(temperatures)
lowest_temp = np.min(temperatures)
hot_days = temperatures[temperatures > 23]
hot_day_count = len(hot_days)

print("Average temperature:", average_temp)
print("Highest temperature:", highest_temp)
print("Lowest temperature:", lowest_temp)
print("Temperatures above 23°C:", hot_days)
print("Number of days above 23°C:", hot_day_count)

assert average_temp == 24.0
assert highest_temp == 28 and lowest_temp == 21
assert list(hot_days) == [25, 28, 24]
assert hot_day_count == 3
````

## Practice Question: Creating a Research Dataset

A scientist records the following information about five plants:

| Plant | Initial Height (cm) | Final Height (cm) | Sunlight (hours) |
|:------|--------------------:|------------------:|-----------------:|
| Rose  | 10 | 18 | 6 |
| Tulip | 12 | 20 | 7 |
| Lily  | 8 | 15 | 5 |
| Rose  | 11 | 17 | 8 |
| Tulip | 14 | 23 | 6 |

1. Create a dictionary containing the given data.
2. Convert the dictionary into a Pandas DataFrame called `df`.
3. Display the first five rows.
4. Display the shape and column names of the DataFrame.

````python
plant_data = {
    'Plant': ['Rose', 'Tulip', 'Lily', 'Rose', 'Tulip'],
    'Initial Height (cm)': [10, 12, 8, 11, 14],
    'Final Height (cm)': [18, 20, 15, 17, 23],
    'Sunlight (hours)': [6, 7, 5, 8, 6]
}

df = pd.DataFrame(plant_data)

print(df.head())
print("Shape:", df.shape)
print("Columns:", list(df.columns))
````

We can convert this dictionary into a DataFrame using `pd.DataFrame()`.

```python
plants_df = pd.DataFrame(plant_data)
```

Here are some useful Pandas methods and attributes:

| Function / Method | What it does |
|:------------------|:-------------|
| `pd.DataFrame()` | Creates a DataFrame from data. |
| `.head()` | Displays the first five rows. |
| `.shape` | Returns the number of rows and columns. |
| `.columns` | Displays the column names. |

Let's explore our DataFrame:

````python
data = {
    'Plant': ['Rose', 'Tulip', 'Lily', 'Rose', 'Tulip'],
    'Initial_Height': [10, 12, 8, 11, 14],
    'Final_Height': [18, 20, 15, 17, 23],
    'Sunlight': [6, 7, 5, 8, 6]
}

plants_df = pd.DataFrame(data)
print(plants_df.head())

print("Shape:", plants_df.shape)

print("Columns:", plants_df.columns)
````

### Selecting and Filtering Data

We can select a particular column by using its name.

We can also filter rows based on a condition.

For example, let's find all plants with a height greater than 12 cm.

````python
# The dataset has no 'Height' column; the height measurements are 'Final_Height'
print(plants_df['Final_Height'])

taller_plants = plants_df[plants_df['Final_Height'] > 12]

print(taller_plants)
````

### Grouping Data

The `groupby()` method allows us to group rows that share the same value.

For example, we can calculate the average height of each plant type.

````python
average_height = plants_df.groupby('Plant')['Final_Height'].mean()

print(average_height)
````

### Question: **Try and explain this bit of code yourself**
```python
plants_df.groupby('Plant')['Final_Height'].mean()
```

*(Note: the original cell used `'Height'`, but our DataFrame has no column of that name, so it raises a `KeyError`. The column holding the plant heights is `Final_Height`, so that is what the code above uses.)*

### Answer here:

This line works in three steps:

1. `plants_df.groupby('Plant')` splits the DataFrame into groups, one for every distinct value in the `Plant` column (Lily, Rose and Tulip). Rows with the same plant name end up in the same group.
2. `['Final_Height']` selects just the height column within each group, so only those values are used in the calculation.
3. `.mean()` calculates the average of those values for each group.

The result is a Series indexed by plant name, giving the average height of each plant type. For example, Rose has final heights 18 and 17, so its average is 17.5, and Tulip has 20 and 23, so its average is 21.5. Lily has only one plant, so its average is just 15.

### Task 2
Continue working with the DataFrame created in the practice question.

- Create a new column called `Growth` by subtracting the initial height from the final height.

- Display only the `Plant` and `Growth` columns.

- Filter and display plants that grew more than 7 cm.

- Calculate the average growth for each plant type using `groupby()`.

````python
plants_df['Growth'] = plants_df['Final_Height'] - plants_df['Initial_Height']

print(plants_df[['Plant', 'Growth']])

print(plants_df[plants_df['Growth'] > 7])

print(plants_df.groupby('Plant')['Growth'].mean())

assert list(plants_df['Growth']) == [8, 8, 7, 6, 9]
````

## Part 4 : Combining everything!
### The School Science Fair

Your school has just held its annual Science Fair! Six students participated, and their marks in Physics, Chemistry and Biology have been recorded.

Your job is to analyse their performance and identify the top-performing students.

### Research Dataset

| Student | Physics | Chemistry | Biology |
|:--------|--------:|-----------:|--------:|
| Asha | 85 | 78 | 92 |
| Jess | 72 | 88 | 80 |
| Meera | 95 | 91 | 89 |
| Rohan | 65 | 70 | 75 |
| Sara | 88 | 84 | 90 |
| Dev | 78 | 76 | 82 |

### Your Mission

**Step 1: Create the Dataset using Pandas**

- Store the information in a dictionary.
- Convert the dictionary into a DataFrame called `students_df`.
- Display the dataset.

**Step 2: Analyse the Scores using NumPy**

- Convert the Physics scores into a NumPy array.
- Calculate the average Physics score.
- Find the highest and lowest Physics scores.

**Step 3: Analyse Student Performance using Pandas**

- Create a new column called `Total` by adding the three subject scores.
- Create another column called `Average` to calculate each student's average marks.
- Filter and display students whose average score is greater than 80.
- Find the student with the highest average score.

**Step 4: Visualise Your Findings using Matplotlib**

- Create a bar graph comparing the average scores of all six students.
- Add a suitable title and label both axes.
- Add a grid to make the graph easier to read.

**Step 5: Answer the Following Questions**

1. Which student scored the highest in Physics?
2. What was the average Physics score?
3. Which student had the highest overall average?
4. How many students scored an average above 80?

````python
# Step 1: Creating a DataFrame from the dictionary
students = {
    'Student': ['Asha', 'Jess', 'Meera', 'Rohan', 'Sara', 'Dev'],
    'Physics': [85, 72, 95, 65, 88, 78],
    'Chemistry': [78, 88, 91, 70, 84, 76],
    'Biology': [92, 80, 89, 75, 90, 82]
}
students_df = pd.DataFrame(students)
print(students_df)

# Step 2: NumPy analysis of the Physics scores
physics = np.array(students_df['Physics'])
avg_physics = np.mean(physics)
max_physics = np.max(physics)
min_physics = np.min(physics)
print("\nAverage Physics score:", avg_physics)
print("Highest Physics score:", max_physics)
print("Lowest Physics score:", min_physics)

# Step 3: Calculating the total marks of each student
students_df['Total'] = students_df['Physics'] + students_df['Chemistry'] + students_df['Biology']

# Calculating the average marks of each student
students_df['Average'] = students_df['Total'] / 3
print()
print(students_df)

# Filtering students with an average above 80
above_80 = students_df[students_df['Average'] > 80]
print("\nStudents with an average above 80:")
print(above_80[['Student', 'Average']])

# Finding the highest average score
best_avg = students_df['Average'].max()
best_student = students_df.loc[students_df['Average'].idxmax(), 'Student']
print("\nHighest average:", best_student, "with", round(best_avg, 2))

# Step 4: Bar graph of average scores
plt.bar(students_df['Student'], students_df['Average'])
plt.title("Average Scores of Science Fair Students")
plt.xlabel("Student")
plt.ylabel("Average Mark")
plt.grid(axis='y')
plt.show()

# Step 5: Answers
top_physics_student = students_df.loc[students_df['Physics'].idxmax(), 'Student']
print("1. Highest in Physics:", top_physics_student, "(", max_physics, ")")
print("2. Average Physics score:", avg_physics)
print("3. Highest overall average:", best_student)
print("4. Students with average above 80:", len(above_80))

assert top_physics_student == 'Meera'
assert best_student == 'Meera'
assert len(above_80) == 3
````
